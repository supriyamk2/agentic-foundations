## Calling LLMs from multple providers

We are about to call LLMs from many other providers.

````python
# OpenAI Compatible URLs

ANTHROPIC_BASE_URL = "https://api.anthropic.com/v1/"
DEEPSEEK_BASE_URL = "https://api.deepseek.com/v1"
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
GROQ_BASE_URL = "https://api.groq.com/openai/v1"
GROK_BASE_URL = "https://api.x.ai/v1"
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"
OLLAMA_BASE_URL = "http://localhost:11434/v1"
````

````python
# OpenAI client libraries with the right base_url and key
# If this surprises you, please see Guide 9 in the Guides folder!

anthropic = OpenAI(api_key=anthropic_api_key, base_url=ANTHROPIC_BASE_URL)
deepseek = OpenAI(api_key=deepseek_api_key, base_url=DEEPSEEK_BASE_URL)
gemini = OpenAI(api_key=google_api_key, base_url=GEMINI_BASE_URL)
groq = OpenAI(api_key=groq_api_key, base_url=GROQ_BASE_URL)
grok = OpenAI(api_key=grok_api_key, base_url=GROK_BASE_URL)
openrouter = OpenAI(api_key=openrouter_api_key, base_url=OPENROUTER_BASE_URL)
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
````

## And now for Grok!

Branded as "The most truth-seeking large language model in the world".. so let's use it as our LLM as a judge

````python
# Judgement time!
# Grok is "The most truth-seeking large language model in the world."

model_name = "claude-opus-5-5"

response = anthropic.chat.completions.create(model=model_name, messages=judge_messages)
results = response.choices[0].message.content
print(results)
````

````python
# OK let's turn this into results!

results_dict = json.loads(results)
print(results_dict)
ranks = results_dict["results"]
print(ranks)
for index, result in enumerate(ranks):
    competitor = competitors[int(result)-1]
    print(f"Rank {index+1}: {competitor}")
````

````python
ranks = results_dict["results"]

worst_index = int(ranks[-1]) - 1

worst_model = competitors[worst_index]
worst_answer = answers[worst_index]

print("Worst model:", worst_model)
print("Original answer:", worst_answer)
````

````python
import json

# Automatically generate competitor IDs
model_ids = [str(i) for i in range(1, len(competitors) + 1)]

# Build the expected JSON structure
output_format = {
    "results": model_ids,
    "feedback": {
        model_id: "Specific feedback for this model"
        for model_id in model_ids
    }
}

judge = f"""
You are evaluating {len(competitors)} LLM responses.

Question:
{question}

Responses:
{together}

Evaluate each response based on:
1. Clarity
2. Accuracy
3. Strength of reasoning

Rank all competitors from best to worst.

Provide specific, actionable feedback for each competitor.

Return ONLY valid JSON matching this structure:

{json.dumps(output_format, indent=2)}

The results array must contain every competitor ID
exactly once, ordered from best to worst.
"""
````

````python
print(json.dumps(output_format, indent=2))
````

````python
judge_messages = [
    {"role": "user", "content": judge}
]

response = anthropic.chat.completions.create(
    model="claude-opus-5-5",
    messages=judge_messages
)

results = response.choices[0].message.content
````

````python
import json

results_dict = json.loads(results)

print(json.dumps(results_dict, indent=2))
````

````python
for model_id, feedback in results_dict["feedback"].items():
    model_name = competitors[int(model_id) - 1]

    print(f"\nModel: {model_name}")
    print(f"Feedback: {feedback}")
````

````python
worst_id = results_dict["results"][-1]

worst_model = competitors[int(worst_id) - 1]
worst_answer = answers[int(worst_id) - 1]
worst_feedback = results_dict["feedback"][worst_id]

print("Weakest model:", worst_model)
print("Original answer:", worst_answer)
print("Judge feedback:", worst_feedback)
````

````python
# Map each model to its corresponding API client
model_clients = {
    "gpt-5.4-nano": openai,
    "claude-sonnet-4-6": anthropic,
    "gemini-3.1-flash-lite": gemini,
    "deepseek-v4-flash": deepseek,
    "openai/gpt-oss-120b": groq,
    "moonshotai/kimi-k2.6": openrouter,
    "llama3.2:1b": ollama,
    "gpt-oss:latest": ollama,
    "gemma4:latest": ollama,
}

reflection_prompt = f"""
Original question:
{question}

Your previous answer:
{worst_answer}

Evaluator feedback:
{worst_feedback}

Reflect on the feedback and improve your answer.
Keep it concise.
Return only the improved answer.
"""

# Get the correct client
client = model_clients[worst_model]

# Send feedback to the original model
response = client.chat.completions.create(
    model=worst_model,
    messages=[
        {"role": "user", "content": reflection_prompt}
    ]
)

improved_answer = response.choices[0].message.content

print("Model:", worst_model)
print("Original answer:", worst_answer)
print("Improved answer:", improved_answer)
````

````python
answers[worst_index] = improved_answer
````

````python
together = ""

for index, answer in enumerate(answers):
    together += f"""
# Response from competitor {index + 1}

{answer}

"""
````

````python
judge = f"""
You are evaluating {len(competitors)} LLM responses.

Question:
{question}

Responses:
{together}

Evaluate each response for:
1. Clarity
2. Accuracy
3. Strength of reasoning

Rank all competitors from best to worst.

Return a JSON object containing:
- "results": an array of all competitor IDs, best to worst
- "feedback": an object containing specific feedback for every competitor

Competitor IDs:
{[str(i) for i in range(1, len(competitors) + 1)]}

IMPORTANT:
Return only a valid JSON object.
Do not include Markdown code fences.
Do not include explanations outside the JSON.
"""

response = anthropic.chat.completions.create(
    model="claude-opus-5-5",
    messages=[
        {"role": "user", "content": judge}
    ]
)

raw_response = response.choices[0].message.content

print(repr(raw_response))
````

````python
import json
import re

raw_response = response.choices[0].message.content

if not raw_response or not raw_response.strip():
    raise ValueError("Judge returned an empty response")

cleaned = raw_response.strip()

# Remove Markdown code fences if present
cleaned = re.sub(r"^```(?:json)?\s*", "", cleaned)
cleaned = re.sub(r"\s*```$", "", cleaned)

new_results = json.loads(cleaned)

print("Original ranking:", ranks)
print("New ranking:", new_results["results"])
````

````python
print("\nUPDATED RANKINGS")

for position, model_id in enumerate(new_results["results"], start=1):
    index = int(model_id) - 1

    print(f"Rank {position}: {competitors[index]}")
    print(f"Feedback: {new_results['feedback'][model_id]}")
````
